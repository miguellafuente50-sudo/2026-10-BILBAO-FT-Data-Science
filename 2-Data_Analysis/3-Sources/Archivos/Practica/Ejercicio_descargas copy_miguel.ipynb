### 1. Crea un un programa con el flujo principal y encapsúlalo en funciones.

```python
import os
import shutil
from pathlib import Path
```

```python
#Crear carpetas 
def crear_capetas(ruta):
    carpetas = ["Imagenes", "Documentos", "Software", "Otros"]
    for carpeta in carpetas:
           os.makedirs(os.path.join(ruta, carpeta), exist_ok=True)

#Identificar el tipo de archivo
def obtener_categoria(archivo):
    extension = os.path.splitext(archivo)[1].lower()
    
    imagenes = [".jpg", ".jpeg", ".png", ".gif", ".svg"]
    documentos = [".pdf", ".doc", ".docx", ".txt", ".xls",
                  ".xlsx", ".ppt", ".pptx"]
    software = [".exe", ".deb", ".zip", ".rar", ".tar", ".gz"]
    if extension in imagenes:
        return "Imagenes"
    elif extension in documentos:
        return "Documentos"
    elif extension in software:
        return "Software"
    else:
        return "Otros"
#Mover archivo  del orgigen a nuetra carpeta creada, segun su categoria
def mover_archivo(ruta, archivo, categoria):

    origen = os.path.join(ruta, archivo)
    destino = os.path.join(ruta, categoria, archivo)

    if not os.path.isfile(origen):
        print(f"No existe el archivo: {archivo}")
        return

    if not os.path.isdir(os.path.join(ruta, categoria)):
        print(f"No existe la carpeta: {categoria}")
        return

    if os.path.exists(destino):
        print(f"Ya existe en destino: {archivo}")
        return

    shutil.move(origen, destino)

    print(f"Movido: {archivo} -> {categoria}")
#Flujo principal
def main():
    ruta_descargas = str(Path.home() / "Descargas" / "Carpeta_copia_ejrecicioç")
    if not os.path.isdir(ruta_descargas):
        print("No se encuentra la Carpeta_copia_ejrecicioç")
        return
    if not os.path.isdir(ruta_descargas):
        print("No se encuentra la Carpeta_copia_ejrecicioç")
        return
    
    print("Organizado  archivos...")

    crear_capetas(ruta_descargas)

    organizar_archivo(ruta_descargas)

    print("Organización finalizada.")
#Ejecutar programa
if __name__ == "__main__":
    main()
```
