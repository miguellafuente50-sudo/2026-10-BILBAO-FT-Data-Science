## Ejercicio 13
Representa el nivel de alcohol (columna `abv`) en un histograma. ¿Se trata de una distribución simétrica o asimétrica?

```python
import matplotlib.pyplot as plt

beers["abv"].dropna().hist(bins=20, edgecolor="black")
plt.xlabel("Alcohol por volumen (ABV)")
plt.ylabel("Número de cervezas")
plt.title("Distribución del nivel de alcohol")
plt.show()
```

## Ejercicio 14
Representa mediante un *boxplot* de `seaborn` la variable `abv`

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(y=beers["abv"])
plt.ylabel("Alcohol por volumen (ABV)")
plt.title("Boxplot del nivel de alcohol")
plt.show()
```

## Ejercicio 24
Teniendo en cuenta la media de amargor de todas las cervezas, obtén una tabla con las cervezas cuyos estados tengan una media de amargor en cerveza por encima de la media

```python
media_global = unidos["ibu"].mean()
print(media_global)
media_estado = unidos.groupby("state")["ibu"].mean()
print(media_estado)
estados_sobre_media = media_estado[media_estado > media_global].index
print(estados_sobre_media)
tabla = unidos[unidos["state"].isin(estados_sobre_media) & unidos["ibu"].notna()
][["name", "style", "ibu", "state"]]

tabla
```

```python
Para encontrar las cervezas de estados cuyo amargor medio supera la media general:

unidos["ibu"].mean() calcula el IBU medio de todas las cervezas.
groupby("state")["ibu"].mean() calcula el IBU medio por estado.
Se seleccionan los estados cuya media es mayor que la global.
Se muestran las cervezas de esos estados, con su nombre, estilo, IBU y estado.
Los IBU vacíos no entran en los cálculos ni aparecen en la tabla.
```
