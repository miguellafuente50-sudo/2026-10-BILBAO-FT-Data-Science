    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo:

* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus

* Resuélvelo con funciones:

```python
#Creamos la función tblero
def crear_tablero(x=10, y=10):
    tablero = []
    for i in range(y):
        fila = []
        for j in range(x):
            fila.append("~")
        tablero.append(fila)

    return tablero

tablero = crear_tablero()
```

```python
import random
#Funció crear barcos fijos
tipos_barco = {
1:4,
2:3,
3:2,
4:3
}

def conseguir_celda_libre(tablero, longitud, direccion):
    fila_maxima = len(tablero)-1
    columna_maxima = len(tablero[0])-1
    if direccion == "Horizontal":
        columna_maxima = len(tablero[0]) - longitud
    if direccion == "Vertical":
        fila_maxima = len(tablero) - longitud

    while True:
        fila = random.randint(0, fila_maxima)
        columna =  random.randint(0,columna_maxima)
        fila_actual = fila
        columna_actual = columna
        ocupado = False
        for i in range(0, longitud):
            if tablero[fila_actual][columna_actual] == "B":
                ocupado = True
                break
            if direccion == "Horizontal":
                columna_actual += 1
            elif direccion == "Vertical":
                fila_actual += 1
        if not ocupado:
            return fila, columna


def crear_barco(tablero, longitud, direccion="Horizontal"):
    fila, columna = conseguir_celda_libre(tablero, longitud, direccion)
    fila_actual = fila
    columna_actual = columna
    for i in range(0, longitud):
        tablero[fila_actual][columna_actual] = "B"
        if direccion == "Horizontal":
            columna_actual += 1
        elif direccion == "Vertical":
            fila_actual += 1


def crear_direccion():
    if random.random() < 0.5:
        return "Vertical"
    else:
        return "Horizontal"



def crear_barcos(tablero,tipos_barco):
    for tipo in tipos_barco:
        longitud = tipo
        for i in range(tipos_barco[tipo]):
            direccion = crear_direccion()
            crear_barco(tablero, longitud, direccion)

tablero = crear_tablero(10,10)
crear_barcos(tablero, tipos_barco)
tablero

#No he podido terminar.
```
