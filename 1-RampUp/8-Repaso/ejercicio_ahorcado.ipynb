## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import random

palabras = ["salsa", "python", "casa", "perro", "gato"]

palabra = random.choice(palabras)
print(palabra)

#Dependiendo de la lonjutud de la palabra, habra una cantidad equivalente de lineas 
rayas = []
for i in range(len(palabra)):
    rayas.append("_")
#print(" ".join(rayas))

#Si la l esta en la palabra se queda la letra
vidas = 6
while vidas > 0 :
    letra_jugador = input("Escribe una letra:")
    if letra_jugador in palabra:
        for l, letra in enumerate(palabra):
            if letra_jugador == letra:
                rayas[l] = letra_jugador
                print(palabra[l])
                print("Has acertado")

                if "_" not in rayas:
                    print("Has ganado")
                    vidas = 0
                
                    

    else:
        vidas -= 1
        print("Pierdes vidas")
        print("Tus vidas son:",vidas)
        print("Vaya paquete, has perdido")
    
    print(" ".join(rayas))


        







````
